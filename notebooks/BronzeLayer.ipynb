### Read All Raw Data

```python
customers_raw = spark.read.parquet("abfss://EcommerceWorkSpce@onelake.dfs.fabric.microsoft.com/EcommerceLakeHouse.Lakehouse/Files/Bronze/customers.parquet")
orders_raw = spark.read.parquet("abfss://EcommerceWorkSpce@onelake.dfs.fabric.microsoft.com/EcommerceLakeHouse.Lakehouse/Files/Bronze/orders.parquet")
payments_raw = spark.read.parquet("abfss://EcommerceWorkSpce@onelake.dfs.fabric.microsoft.com/EcommerceLakeHouse.Lakehouse/Files/Bronze/payments.parquet")
support_tickets_raw = spark.read.parquet("abfss://EcommerceWorkSpce@onelake.dfs.fabric.microsoft.com/EcommerceLakeHouse.Lakehouse/Files/Bronze/support_tickets.parquet")
web_activities_raw = spark.read.parquet("abfss://EcommerceWorkSpce@onelake.dfs.fabric.microsoft.com/EcommerceLakeHouse.Lakehouse/Files/Bronze/web_activities.parquet")
```

### Write Data To Delta Table

```python
customers_raw.write.format('delta').mode('overwrite').saveAsTable('customers')
orders_raw.write.format('delta').mode('overwrite').saveAsTable('orders')
payments_raw.write.format('delta').mode('overwrite').saveAsTable('payments')
support_tickets_raw.write.format('delta').mode('overwrite').saveAsTable('support')
web_activities_raw.write.format('delta').mode('overwrite').saveAsTable('web')
```
